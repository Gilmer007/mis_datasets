<h2><font color="#004D7F" size=6>IDENTIFICACION DE OUTLIERS PUNTUALES POR METODO DE DESVIACION ESTANDAR</font></h2>

```python
#IDENTIFICACION DE OUTLIERS PUNTUALES POR METODO DE DESVIACION ESTANDAR

import warnings
warnings.filterwarnings('ignore')

#CARGAR LIBRERIAS
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

#LEER LOS DATOS
df = pd.read_csv('E:/UNAJ/maes_cd/complemento_cd/R1_ING_IND-copia.csv',header=0)

#CALCULAR MEDIA
media = df['PG_ICFES'].mean()
print("La media de la columna PG_ICFES es:", media)

#CALCULAR DESVIACION ESTANDAR
std = df['PG_ICFES'].std()
print("La desviacion estandar es:", std)

#DEFINIR EL CORTE
corte = std * 3

#CALCULAR RANGOS
Rango_inferior = media - corte
Rango_superior = media + corte

print('El rango inferior es:', Rango_inferior)
print('El rango superior es:', Rango_superior)

#TAMAÑO DEL DATASET
print(df.shape)
print('_____')

#IDENTIFICAR OUTLIERS
outliers = [x for x in df['PG_ICFES'] if x < Rango_inferior or x > Rango_superior]

print('Outliers identificados en la columna PG_ICFES:', len(outliers))

#IDENTIFICAR DATOS NORMALES
outliers_removed = [x for x in df['PG_ICFES'] if x >= Rango_inferior and x <= Rango_superior]

print('Observaciones que no son outliers en la columna PG_ICFES:', len(outliers_removed))
```

<h2><font color="#004D7F" size=6>Calcular rango intercuartílico</font></h2>

```python
# CONVERTIR LA COLUMNA A NUMERICA
df['PG_ICFES'] = pd.to_numeric(df['PG_ICFES'], errors='coerce')

# MOSTRAR CANTIDAD DE VALORES NULOS
print(df['PG_ICFES'].isnull().sum())

# ELIMINAR VALORES NULOS
df = df.dropna(subset=['PG_ICFES'])

# IMPORTAR LIBRERIA
from numpy import percentile

# CALCULAR PERCENTILES
P_25 = percentile(df['PG_ICFES'], 25)
P_75 = percentile(df['PG_ICFES'], 75)

# CALCULAR RANGO INTERCUARTILICO
Rango_IQR = P_75 - P_25

print('Percentil 25 =', P_25)
print('Percentil 75 =', P_75)
print('Rango IQR =', Rango_IQR)

# CALCULAR LIMITES
corte = Rango_IQR * 1.5

Limite_inferior = P_25 - corte
Limite_superior = P_75 + corte

print('Límite inferior =', Limite_inferior)
print('Límite superior =', Limite_superior)

# ELIMINAR OUTLIERS
df = df[(df['PG_ICFES'] >= Limite_inferior) &
        (df['PG_ICFES'] <= Limite_superior)]

# MOSTRAR RESULTADO FINAL
print("Datos después de eliminar outliers:", df.shape)
```
